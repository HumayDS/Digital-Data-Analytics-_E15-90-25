## 6. Outliers: the IQR method
An **outlier** is a value that is unusually high or low compared with the rest of the data.

- **IQR** (Interquartile Range) = Q3 − Q1
- **Lower bound** = Q1 − 1.5 × IQR
- **Upper bound** = Q3 + 1.5 × IQR

Values that fall outside these bounds are outliers.

```python
df_tip = df_1["tip"]
df_tip.describe()
```

```python
# Quartiles, IQR and bounds
Q1 = df_tip.quantile(0.25)
Q3 = df_tip.quantile(0.75)
IQR = Q3 - Q1

below = Q1 - 1.5 * IQR
above = Q3 + 1.5 * IQR
below, above
```

```python
# Rows whose tip is an outlier
df_1[df_1["tip"] > above]
```

```python
# Keep only values within the bounds and renumber the rows
clear_df = df_1[df_1["tip"].between(below, above)].reset_index(drop=True)
clear_df['tip'].describe()
```

```python
# Import the required libraries
import numpy as np
import pandas as pd
import statsmodels.api as sm
from sklearn.metrics import mean_squared_error, r2_score

# Define the independent variables (X) and the dependent variable (y)
X = clear_df.drop('tip', axis=1)  # select all columns except tip
y = clear_df['tip']               # Target variable

# Convert categorical variables to numeric form
X = pd.get_dummies(X, drop_first=True, dtype=int)

# Add an intercept (constant) to the model
X = sm.add_constant(X)

# Create and fit the linear regression model
model = sm.OLS(y, X).fit()

# Compute the predictions
y_pred = model.predict(X)

# SSE — Sum of Squared Errors
sse = np.sum((y - y_pred) ** 2)

# MSE — Mean Squared Error
mse = mean_squared_error(y, y_pred)

# RMSE — square root of the Mean Squared Error
rmse = np.sqrt(mse)

# Print the results
print("Regression model results:")
print(model.summary())

# Print the error metrics
print("\nError metrics:")
print("SSE:", sse)
print("MSE:", mse)
print("RMSE:", rmse)
print("R²:", model.rsquared)

# Show each variable's coefficient, p-value and confidence interval
results = pd.DataFrame({
    'Coefficient': model.params,
    'Std. error': model.bse,
    't-statistic': model.tvalues,
    'P-value': model.pvalues,
    '95% CI lower': model.conf_int()[0],
    '95% CI upper': model.conf_int()[1]
})

print("\nStatistical results by variable:")
print(results.round(4))

# Check statistical significance based on the p-value
print("\nStatistical significance of the variables:")

for variable, p_value in model.pvalues.items():
    if variable == 'const':
        continue

    if p_value < 0.05:
        print(f"{variable}: Statistically significant (p-value = {p_value:.4f})")
    else:
        print(f"{variable}: Not statistically significant (p-value = {p_value:.4f})")
```
